```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
from sklearn.datasets import load_iris
#load the iris dataset and create a DataFrame
iris = load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target
#1.Univariate visualizations
#Histogram
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True)
plt.title('Distribution of Sepal Length')       
plt.show()

#Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title('Box Plot of Sepal Length')
plt.show()

#Pie Chart
species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts, labels=iris.target_names, autopct='%1.1f%%')
plt.title('Distribution of Species')
plt.show()
```

```python
#Bivariate visualizations
#Scatter Plot

plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'], palette='Set1')
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

```python
#Line Chart
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title('Trend of Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
#Bar Chart (mean sepal length by species)
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Mean Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
plt.show()
```

```python
#Multivariate visualizations
#HeatMap (correlation matrix)

plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Iris Features')
plt.show()

# bubble chart (sepal length vs sepal width with species as size and color)
plt.figure(figsize=(6,4))
sns.scatterplot(
    x=df['sepal length (cm)'],
    y=df['sepal width (cm)'],
    size=df['petal length (cm)'] * 20,
    hue=df['species'],
    alpha=0.5
)
plt.title('Sepal Length vs Sepal Width with Petal Length as Size')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()

#Pair plot
sns.pairplot(df.iloc[:,:4])
plt.suptitle('Pair Plot of Iris Features', y=1.02)
plt.show()
```

```python
#Advanced visualizations
#Violin Plot

plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Violin Plot of Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# load the tips dataset
df = sns.load_dataset('tips')

# 1. Univariate visualizations

# Histogram
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title('Distribution of Total Bill')
plt.show()

# Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['total_bill'])
plt.title('Box Plot of Total Bill')
plt.show()

# Pie Chart
day_counts = df['day'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title('Distribution of Days')
plt.show()


# 2. Bivariate visualizations

# Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(
    x=df['total_bill'],
    y=df['tip'],
    hue=df['sex'],
    palette='Set1'
)
plt.title('Total Bill vs Tip')
plt.show()

# Line Chart
plt.figure(figsize=(6,4))
plt.plot(df['total_bill'])
plt.title('Trend of Total Bill')
plt.xlabel('Sample Index')
plt.ylabel('Total Bill')
plt.show()

# Bar Chart (mean total bill by day)
plt.figure(figsize=(6,4))
sns.barplot(x=df['day'], y=df['total_bill'])
plt.title('Mean Total Bill by Day')
plt.show()


# 3. Multivariate visualizations

# HeatMap (correlation matrix)
plt.figure(figsize=(8,6))
sns.heatmap(df[['total_bill', 'tip', 'size']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Tips Features')
plt.show()

# Bubble Chart (total bill vs tip with size and color)
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'],y=df['tip'],size=df['size'] * 20,hue=df['sex'],alpha=0.5)
plt.title('Total Bill vs Tip with Party Size as Size')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()

# Pair Plot
sns.pairplot(df[['total_bill', 'tip', 'size']])
plt.suptitle('Pair Plot of Tips Features', y=1.02)
plt.show()


# 4. Advanced visualizations

# Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['day'], y=df['total_bill'])
plt.title('Violin Plot of Total Bill by Day')
plt.show()
```
