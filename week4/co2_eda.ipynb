```python
import matplotlib.pyplot as pyplot
y1=[]
y2=[]
x=range(-100, 100, 10)
for i in x:
    y1.append(i**2)
for i in x:
    y2.append(-i**2)
pyplot.plot(x,y1)
pyplot.plot(x,y2)
pyplot.xlabel("x axis")
pyplot.ylabel("y axis")
pyplot.title("Simple graph")
pyplot.axhline()
pyplot.axvline()
pyplot.show()
```

```python
import matplotlib.pyplot as pyplot
y1=[]
y2=[]
x=range(-100, 100, 10)
for i in x:
    y1.append(i**2)
for i in x:
    y2.append(-i**2)
pyplot.plot(x,y1, color='Maroon')
pyplot.plot(x,y2,color='Teal')
pyplot.xlabel("x axis")
pyplot.ylabel("y axis")
pyplot.title("Simple graph")
pyplot.axhline(color='red')
pyplot.axvline(color='green')
pyplot.savefig("linegraph.jpg")
pyplot.grid(True)
pyplot.show()
```
